```python
import pandas as pd
import numpy as np
from IPython.display import display

df = pd.read_csv("../data/Superstore.csv", encoding="latin1")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
print("Column names:", df.columns.tolist())

display(df.head())
```

```python
print("Total rows and columns:", df.shape)
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
```

```python
missing_values = df.isnull().sum()

print(missing_values)
print("Total missing values:", missing_values.sum())
```

```python
print("Complete duplicate rows:", df.duplicated().sum())

print(
    "Duplicate Row IDs:",
    df["Row ID"].duplicated().sum()
)
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], errors="coerce"
)

print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
print(
    "Invalid Order Dates:",
    df["Order Date"].isna().sum()
)

print(
    "Invalid Ship Dates:",
    df["Ship Date"].isna().sum()
)
```

```python
df[["Sales", "Quantity", "Discount", "Profit"]].describe()
```

```python
print("Sales <= 0:", (df["Sales"] <= 0).sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("Discount < 0:", (df["Discount"] < 0).sum())
print("Discount > 1:", (df["Discount"] > 1).sum())
```

```python
df.to_csv(
    "../data/cleaned_superstore.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
# Check duplicate records
print("Complete duplicate rows:", df.duplicated().sum())
print("Duplicate Row IDs:", df["Row ID"].duplicated().sum())

# Convert dates
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")

# Check invalid dates
print("Invalid Order Dates:", df["Order Date"].isna().sum())
print("Invalid Ship Dates:", df["Ship Date"].isna().sum())
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()

profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
print("Distinct Orders:", total_orders)
print("Distinct Customers:", total_customers)
print("Profit Margin (%):", round(profit_margin, 2))
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)
```

```python
import matplotlib.pyplot as plt
```

```python
category_sales.plot(
    kind="bar",
    title="Sales by Category"
)

plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(region_profit)
```

```python
region_profit.plot(
    kind="bar",
    title="Profit by Region"
)

plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_products)
```

```python
top_products.sort_values().plot(
    kind="barh",
    title="Top 10 Products by Sales",
    figsize=(10, 6)
)

plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python
monthly_sales = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Sales"]
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

monthly_sales.plot(
    figsize=(12, 5),
    title="Monthly Sales Trend"
)

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(category_profit)

category_profit.plot(
    kind="bar",
    title="Profit by Category",
    figsize=(8, 5)
)

plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
discount_analysis = (
    df.groupby("Discount")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Order_Lines=("Order ID", "size")
    )
    .sort_index()
)

display(discount_analysis)
```

```python
discount_analysis["Total_Profit"].plot(
    kind="bar",
    title="Total Profit by Discount Level",
    figsize=(9, 5)
)

plt.xlabel("Discount")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python
monthly_summary = (
    df.groupby(df["Order Date"].dt.to_period("M"))
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .sort_index()
)

print("Top 5 sales months:")
display(
    monthly_summary.nlargest(5, "Total_Sales")
)

print("Bottom 5 sales months:")
display(
    monthly_summary.nsmallest(5, "Total_Sales")
)
```
